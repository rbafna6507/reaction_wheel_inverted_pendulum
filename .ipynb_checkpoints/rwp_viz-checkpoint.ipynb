# Reaction Wheel Pendulum — Visualization Kit

**Instructor-built (the Phase-8 lane).** This notebook contains **only** rendering logic — no physics, no ODE solving, no control. Your simulation produces arrays; this notebook draws them.

## The contract — what you hand me

Everything goes through `make_trajectory(...)`, which normalizes your sim's output into the dict all the plot functions eat:

| argument | required? | meaning |
|---|---|---|
| `t` | yes | 1-D array of times, seconds (any sample rate — plots/animation resample as needed) |
| `theta` **or** `wheel_x, wheel_y` | one of them | arm angle (rad) **or** wheel-center position (m). If you hand x/y, θ is recovered via `atan2` + unwrap |
| `phi` or `phi_dot` | optional | wheel angle (rad, world frame) or wheel speed (rad/s — integrated so the spokes can spin) |
| `tau` | optional | torque (N·m) — drawn as an arc-arrow on the rig + its own strip chart |
| `theta_dot` | optional | computed by finite difference if absent (used for the phase portrait) |
| `extras` | optional | dict `{name: array}` — **every entry gets its own strip chart** (motor signal, PID terms, energy, anything) |

**Conventions** (matching the course notes): θ = 0 is upright; θ and φ share the same positive sense; pivot at the origin, y up. On screen, positive θ tips the arm to the **right** (matching Fig. 1 of the notes) — pass `RigStyle(mirror=-1)` if you prefer the other sense. This is a display choice only; the numbers are untouched.

**On your motor-signal plan:** inputs = θ + motor signal, with a motor model converting signal → torque, is a sound architecture — but that conversion is plant physics, i.e. *your* lane. Hand this notebook whatever τ your motor model outputs, and/or the raw signal as an `extras` channel (it will get its own chart so you can see signal and torque side by side).

## The four functions

- `make_trajectory(...)` → normalized dict (do this first, always)
- `snapshot(traj, time=...)` → still frame of the rig at one instant + strip charts with a time cursor (cheap sanity check)
- `dashboard(traj)` → full strip-chart stack + θ–θ̇ phase portrait
- `animate(traj, fps=30, speed=1.0)` → inline replay animation (rig + synced charts). `speed=2` plays twice real time; `save_path="run.mp4"` also writes a video if ffmpeg is installed.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from matplotlib.patches import Circle
from matplotlib.gridspec import GridSpec
from dataclasses import dataclass

# palette — matches the course-notes "engineering pad"
INK, PAPER, GROUND = "#23291F", "#F7FAF3", "#E8EFE2"
GREEN, RED, GRAY   = "#1F5F3F", "#B03A2E", "#55604C"

plt.rcParams.update({
    "figure.facecolor": GROUND, "axes.facecolor": PAPER,
    "axes.edgecolor": INK, "text.color": INK, "axes.labelcolor": INK,
    "xtick.color": INK, "ytick.color": INK, "font.size": 10,
    "animation.embed_limit": 60,   # MB, for inline jshtml animations
})

@dataclass
class RigStyle:
    arm_length:   float = 0.15   # m, pivot -> wheel center (display geometry only)
    wheel_radius: float = 0.05   # m
    n_spokes:     int   = 4
    tau_ref:      float = 0.10   # torque drawn as a full-size arc (N·m)
    mirror:       int   = 1      # +1: positive θ tips right (Fig. 1); -1 flips the display
```

```python
def wheel_xy_from_theta(theta, L=0.15, mirror=1):
    # θ measured from upright; pivot at origin, y up
    theta = np.asarray(theta, float)
    return mirror * L * np.sin(theta), L * np.cos(theta)

def theta_from_wheel_xy(x, y, mirror=1):
    # inverse of the above; returns radians in (-pi, pi] before unwrapping
    return np.arctan2(mirror * np.asarray(x, float), np.asarray(y, float))

def make_trajectory(t, theta=None, wheel_x=None, wheel_y=None, phi=None, phi_dot=None,
                    tau=None, theta_dot=None, extras=None, style=None):
    # Normalize whatever the simulation produced into the dict the plot functions eat.
    style = style or RigStyle()
    t = np.asarray(t, float)
    if t.ndim != 1 or len(t) < 2:
        raise ValueError("t must be a 1-D array with at least 2 samples")

    def chk(name, arr):
        if arr is None:
            return None
        arr = np.asarray(arr, float)
        if arr.shape != t.shape:
            raise ValueError(f"{name} has shape {arr.shape}, t has {t.shape}")
        return arr

    if theta is None:
        wheel_x, wheel_y = chk("wheel_x", wheel_x), chk("wheel_y", wheel_y)
        if wheel_x is None or wheel_y is None:
            raise ValueError("provide theta, or both wheel_x and wheel_y")
        theta = np.unwrap(theta_from_wheel_xy(wheel_x, wheel_y, style.mirror))
    theta   = chk("theta", theta)
    phi     = chk("phi", phi)
    phi_dot = chk("phi_dot", phi_dot)
    tau     = chk("tau", tau)
    theta_dot = chk("theta_dot", theta_dot)
    if theta_dot is None:
        theta_dot = np.gradient(theta, t)
    if phi is None:
        if phi_dot is None:
            phi = np.zeros_like(t)
        else:  # trapezoid-integrate speed so the spokes can spin
            phi = np.concatenate([[0.0], np.cumsum(0.5 * (phi_dot[1:] + phi_dot[:-1]) * np.diff(t))])
    if phi_dot is None:
        phi_dot = np.gradient(phi, t)
    extras = {k: chk(k, v) for k, v in (extras or {}).items()}
    return dict(t=t, theta=theta, theta_dot=theta_dot, phi=phi, phi_dot=phi_dot,
                tau=tau, extras=extras, style=style)
```

```python
def _dir(a, mirror=1):
    # unit vector at angle a from upright, in the display's positive sense
    a = np.asarray(a, float)
    return np.stack([mirror * np.sin(a), np.cos(a)], axis=-1)

def _arc_pts(center, R, a0, a1, mirror=1, n=40):
    a = np.linspace(a0, a1, n)
    return np.asarray(center) + R * _dir(a, mirror)

def draw_rig(ax, theta, phi, tau, style=None, readout=None):
    # Draw one instant of the machine on the given axes.
    s = style or RigStyle()
    L, rw, m = s.arm_length, s.wheel_radius, s.mirror
    ax.set_aspect("equal")
    ax.set_xlim(-1.55 * L, 1.55 * L)
    ax.set_ylim(-0.45 * L, 1.58 * L)
    ax.axis("off")

    # ground + stand
    g = -0.30 * L
    ax.plot([-0.9 * L, 0.9 * L], [g, g], color=INK, lw=2)
    for gx in np.linspace(-0.85 * L, 0.85 * L, 12):
        ax.plot([gx, gx - 0.05 * L], [g, g - 0.07 * L], color=INK, lw=1)
    ax.plot([0, -0.10 * L], [0, g], color=INK, lw=2)
    ax.plot([0,  0.10 * L], [0, g], color=INK, lw=2)

    # upright reference
    ax.plot([0, 0], [0, 1.45 * L], color=GRAY, lw=1, ls=(0, (4, 4)))

    # arm + wheel + spokes + rim marker
    c = (L * _dir(theta, m)).ravel()
    ax.plot([0, c[0]], [0, c[1]], color=INK, lw=5, solid_capstyle="round")
    ax.add_patch(Circle(tuple(c), rw, facecolor=PAPER, edgecolor=INK, lw=2.5, zorder=3))
    for k in range(s.n_spokes):
        e = c + 0.92 * rw * _dir(phi + 2 * np.pi * k / s.n_spokes, m).ravel()
        ax.plot([c[0], e[0]], [c[1], e[1]], color=INK, lw=1.4, zorder=4)
    dot = c + 0.76 * rw * _dir(phi, m).ravel()
    ax.plot(dot[0], dot[1], marker="o", ms=5, color=RED, zorder=5)
    ax.add_patch(Circle(tuple(c), 0.12 * rw, color=INK, zorder=5))
    ax.add_patch(Circle((0, 0), 0.05 * L, facecolor=PAPER, edgecolor=INK, lw=2, zorder=6))

    # θ arc near the pivot
    if abs(theta) > 0.02:
        pts = _arc_pts((0, 0), 0.42 * L, 0.0, theta, m)
        ax.plot(pts[:, 0], pts[:, 1], color=GRAY, lw=1.4)

    # torque arc-arrow around the wheel (length ~ |τ|, sense = sign)
    if tau is not None and abs(tau) > 1e-9:
        frac = min(abs(tau) / s.tau_ref, 1.0)
        sweep = (0.35 + 1.55 * frac) * np.sign(tau)
        pts = _arc_pts(c, 1.55 * rw, 0.0, sweep, m)
        col = GREEN if tau > 0 else RED
        ax.plot(pts[:, 0], pts[:, 1], color=col, lw=2.2, zorder=4)
        ax.annotate("", xy=tuple(pts[-1]), xytext=tuple(pts[-2]),
                    arrowprops=dict(arrowstyle="-|>", color=col, lw=2.2), zorder=4)

    if readout:
        ax.text(0.02, 0.99, readout, transform=ax.transAxes, va="top", ha="left",
                family="monospace", fontsize=9, color=INK)

def _channels(traj):
    ch = [("θ  arm (deg)", np.degrees(traj["theta"]), INK),
          ("φ̇  wheel (rad/s)", traj["phi_dot"], GREEN)]
    if traj["tau"] is not None:
        ch.append(("τ  torque (N·m)", traj["tau"], RED))
    for name, arr in traj["extras"].items():
        ch.append((name, arr, GRAY))
    return ch

def _frame_layout(traj, show_charts=True):
    # figure with the rig on the left and time-cursor strip charts on the right
    ch = _channels(traj)
    if not show_charts:
        fig = plt.figure(figsize=(6, 6))
        return fig, fig.add_subplot(111), []
    fig = plt.figure(figsize=(11, 6))
    gs = GridSpec(len(ch), 2, width_ratios=[1.25, 1], figure=fig, hspace=0.55, wspace=0.25)
    rig_ax = fig.add_subplot(gs[:, 0])
    cursors = []
    t = traj["t"]
    for i, (name, arr, col) in enumerate(ch):
        ax = fig.add_subplot(gs[i, 1])
        ax.plot(t, arr, color=col, lw=1.2)
        ax.axhline(0, color=GRAY, lw=0.6)
        ax.set_ylabel(name, fontsize=8)
        ax.tick_params(labelsize=8)
        if i < len(ch) - 1:
            ax.set_xticklabels([])
        else:
            ax.set_xlabel("time (s)", fontsize=9)
        cursors.append(ax.axvline(t[0], color=RED, lw=1))
    return fig, rig_ax, cursors

def _readout(tv, th, pd, tau_v):
    ro = f"t  = {tv:6.2f} s\nθ  = {np.degrees(th):8.2f}°\nφ̇ = {pd:8.1f} rad/s ({pd*9.5493:6.0f} rpm)"
    if tau_v is not None:
        ro += f"\nτ  = {tau_v:+.3f} N·m"
    return ro

def snapshot(traj, time=None, show_charts=True):
    # Still frame at one instant — the cheap way to sanity-check geometry.
    t = traj["t"]
    tv = float(t[0] if time is None else np.clip(time, t[0], t[-1]))
    at = lambda key: float(np.interp(tv, t, traj[key])) if traj[key] is not None else None
    fig, rig_ax, cursors = _frame_layout(traj, show_charts)
    draw_rig(rig_ax, at("theta"), at("phi"), at("tau"), traj["style"],
             readout=_readout(tv, at("theta"), at("phi_dot"), at("tau")))
    for c in cursors:
        c.set_xdata([tv, tv])
    return fig
```

```python
def dashboard(traj, phase=True, figsize=None):
    # Strip-chart stack for a whole run, plus an optional θ–θ̇ phase portrait.
    ch = _channels(traj)
    n = len(ch) + (1 if phase else 0)
    fig, axes = plt.subplots(n, 1, figsize=figsize or (9, 1.9 * n))
    axes = np.atleast_1d(axes)
    t = traj["t"]
    for ax, (name, arr, col) in zip(axes, ch):
        ax.plot(t, arr, color=col, lw=1.4)
        ax.axhline(0, color=GRAY, lw=0.6)
        ax.set_ylabel(name, fontsize=9)
    axes[len(ch) - 1].set_xlabel("time (s)")
    if phase:
        axp = axes[-1]
        th, thd = np.degrees(traj["theta"]), np.degrees(traj["theta_dot"])
        axp.plot(th, thd, color=INK, lw=1.2)
        axp.plot(th[0], thd[0], "o", color=GREEN, ms=7, label="start")
        axp.plot(th[-1], thd[-1], "o", color=RED, ms=7, label="end")
        axp.axhline(0, color=GRAY, lw=0.6); axp.axvline(0, color=GRAY, lw=0.6)
        axp.set_xlabel("θ (deg)"); axp.set_ylabel("θ̇ (deg/s)")
        axp.legend(fontsize=8, loc="best")
    fig.tight_layout()
    return fig
```

```python
_LAST_ANIM = None   # keep a reference so the animation isn't garbage-collected

def animate(traj, fps=30, speed=1.0, max_frames=400, show_charts=True, save_path=None, dpi=90):
    # Replay a finished run: rig animation + synced time cursors on the charts.
    global _LAST_ANIM
    t = traj["t"]
    t0, t1 = float(t[0]), float(t[-1])
    nf = int(min(max_frames, max(2, round((t1 - t0) / speed * fps))))
    ft = np.linspace(t0, t1, nf)
    at = lambda arr: np.interp(ft, t, arr)
    TH, PH, PD = at(traj["theta"]), at(traj["phi"]), at(traj["phi_dot"])
    TAU = at(traj["tau"]) if traj["tau"] is not None else None

    fig, rig_ax, cursors = _frame_layout(traj, show_charts)

    def update(i):
        rig_ax.clear()
        tau_v = None if TAU is None else float(TAU[i])
        draw_rig(rig_ax, float(TH[i]), float(PH[i]), tau_v, traj["style"],
                 readout=_readout(ft[i], float(TH[i]), float(PD[i]), tau_v))
        for c in cursors:
            c.set_xdata([ft[i], ft[i]])
        return []

    anim = animation.FuncAnimation(fig, update, frames=nf, interval=1000 / fps, blit=False)
    _LAST_ANIM = anim
    if save_path:
        try:
            anim.save(save_path, writer=animation.FFMpegWriter(fps=fps), dpi=dpi)
            print("saved", save_path)
        except Exception as e:
            print("mp4 save failed (is ffmpeg installed?):", e)
    html = None
    try:
        from IPython.display import HTML
        html = HTML(anim.to_jshtml(fps=fps, default_mode="once"))
    except ImportError:
        pass
    plt.close(fig)
    return html if html is not None else anim
```

## ⚠️ SYNTHETIC DEMO — NOT PHYSICS

The cells below feed the renderer a **hand-drawn, made-up motion** purely so you can see it work.
Nothing here solves any equation. **Delete these cells** once your simulation produces real trajectories,
and call the same three functions on your own arrays.

```python
# fake trajectory: a decaying wobble, an invented wheel-speed trace, an invented torque
t_demo   = np.linspace(0.0, 6.0, 1200)
theta_d  = 0.35 * np.exp(-t_demo / 2.2) * np.cos(2 * np.pi * 1.1 * t_demo)
phidot_d = 25.0 * np.exp(-t_demo / 2.2) * np.sin(2 * np.pi * 1.1 * t_demo) + 6.0 * t_demo / 6.0
tau_d    = 0.08 * np.exp(-t_demo / 2.2) * np.cos(2 * np.pi * 1.1 * t_demo + 0.6)

demo = make_trajectory(t_demo, theta=theta_d, phi_dot=phidot_d, tau=tau_d,
                       extras={"motor signal (fake)": np.tanh(tau_d / 0.05)})

snapshot(demo, time=0.4);
```

```python
dashboard(demo);
```

```python
# inline replay — drag the slider / press play. speed=1.0 is real time.
animate(demo, fps=30, speed=1.0)
```
